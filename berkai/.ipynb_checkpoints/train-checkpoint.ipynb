```python
from unsloth import FastLanguageModel
import torch

max_seq_length = 2048 # Ne kadar uzunlukta metinleri hatırlayacağı
dtype = None # Otomatik belirler
load_in_4bit = True # 12GB VRAM'e sığması için 4-bit (QLoRA) kullanıyoruz

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name = "unsloth/llama-3-8b-Instruct-bnb-4bit",
    max_seq_length = max_seq_length,
    dtype = dtype,
    load_in_4bit = load_in_4bit,
)

# Peft (QLoRA) ayarları: Modelin sadece küçük bir kısmını güncelleyeceğiz
model = FastLanguageModel.get_peft_model(
    model,
    r = 16, # LoRA rank, genelde 8, 16 veya 32 olur
    target_modules = ["q_proj", "k_proj", "v_proj", "o_proj", "gate_proj", "up_proj", "down_proj"],
    lora_alpha = 16,
    lora_dropout = 0, # Unsloth 0 öneriyor
    bias = "none",
    use_gradient_checkpointing = "unsloth",
)
```
